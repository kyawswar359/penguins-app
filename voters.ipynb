```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("RestaurantBillsPractice")
    .getOrCreate()
)
```

```python
voters_df = spark.read.json("voters-5K.json")
```

```python
voters_df.printSchema()
```

```python
print("Total records", voters_df.count())
voters_df.show(5)
```

```python
voters_df.groupBy("party").count().orderBy("party").show()
```

```python
#60 Years and older
old_voters = voters_df.filter(F.col("age") >=60)

print("Age 60 and above: ", old_voters.count())
old_voters.show(5)
```

```python
# average age per party
voters_df.groupBy("party").agg(F.round(F.avg("age"), 2).alias("average age")).orderBy("party").show()
```

```python
voters_df.groupBy("gender", "party").count().orderBy("gender", "party").show()
```
